```python
import numpy as np

# Given array - Week 1 temperatures in Celsius
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Original Array (Celsius):")
print(temperatures_w1)

# --- 1. Convert Celsius to Fahrenheit ---
# Formula: Fahrenheit = (Celsius * 9/5) + 32
# NumPy does this for the whole array at once (Vectorization)

temperatures_f = (temperatures_w1 * 9/5) + 32

print("\nConverted Array (Fahrenheit):")
print(temperatures_f)
print(f"Formatted: {np.round(temperatures_f, 2)}")
```

```python
import numpy as np

# Given array - Week 1
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

# Index mapping for clarity:
# Day: Mon Tue Wed Thu Fri Sat Sun
# Index: 0 1 2 3 4 5 6
# Value: 22.5 25.3 20.8 23.4 26.1 24.8 21.9

print(f"Full Week: {temperatures_w1}")

# --- 1. Extract temperatures for the first three days of the week ---
# Days: Monday, Tuesday, Wednesday -> Index 0, 1, 2
# Slicing: [0:3] means start at 0, go up to (but not including) 3

first_three_days = temperatures_w1[0:3]
# Alternative short form: temperatures_w1[:3]

print("\n1. First three days (Mon-Wed):")
print(first_three_days)
```

```python
# --- 3. Extract temperatures for the middle three days ---
# Days: Wednesday, Thursday, Friday -> Index 2, 3, 4
# Slicing: [2:5]

middle_three_days = temperatures_w1[2:5]

print("\n3. Middle three days (Wed-Thu-Fri):")
print(middle_three_days)
```

```python
import pandas as pd

# Create Series
marks = pd.Series([95, 92, 89, 85, 80],
                  index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])

print("Original Series:")
print(marks)

# --- 1. Use integer index position to access mark of 1st rank student ---
# 1st rank is at position 0 (or index label 'Rank1')
first_rank_mark_pos = marks.iloc[0] # FIXED: was marks[0]
print("\n1. Mark of 1st rank student (integer position):")
print(f"marks[0] = {first_rank_mark_pos}")

# --- 2. Use loc accessor to retrieve marks of top 3 ranks by index labels ---
# loc uses LABELS -> 'Rank1' to 'Rank3'
top_3_ranks = marks.loc['Rank1':'Rank3']
# Alternative: marks.loc[['Rank1', 'Rank2', 'Rank3']]

print("\n2. Top 3 ranks using.loc (by labels):")
print(top_3_ranks)

# --- 3. Use iloc accessor to retrieve mark of 3rd-ranked student ---
# iloc uses POSITION -> 3rd rank is at position 2
third_rank_mark = marks.iloc[2]

print("\n3. Mark of 3rd-ranked student using.iloc[2]:")
print(third_rank_mark)

# --- 4. Boolean mask to filter marks > 90 ---
# Create a boolean mask
mask = marks[marks > 90]
print("\n4. Boolean Mask (marks > 90):")
print(mask)

filtered_marks = marks[marks > 90]

print("\nFiltered Ranks where marks > 90:")
print(filtered_marks)
```
